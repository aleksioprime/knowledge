# Линейные структуры данных


Связные списки, стек, очередь, дек — самые базовые структуры.


## Односвязный список (Singly Linked List)

Каждый узел хранит значение и ссылку на следующий узел.

Сложность:
- вставка/удаление в начало — `O(1)`
- вставка/удаление в конец — `O(n)` (нужно пройти до конца) или `O(1)` с хранением `tail`
- поиск/доступ по индексу — `O(n)`


```python
class Node:
    def __init__(self, value):
        self.value = value
        self.next = None

class LinkedList:
    def __init__(self):
        self.head = None

    def append(self, value):
        node = Node(value)
        if self.head is None:
            self.head = node
            return
        current = self.head
        while current.next:
            current = current.next
        current.next = node

    def prepend(self, value):
        node = Node(value)
        node.next = self.head
        self.head = node

    def remove(self, value):
        if self.head is None:
            return
        if self.head.value == value:
            self.head = self.head.next
            return
        current = self.head
        while current.next:
            if current.next.value == value:
                current.next = current.next.next
                return
            current = current.next

    def to_list(self):
        result, cur = [], self.head
        while cur:
            result.append(cur.value); cur = cur.next
        return result

ll = LinkedList()
for x in [1, 2, 3]:
    ll.append(x)
ll.prepend(0)
ll.remove(2)
ll.to_list()   # [0, 1, 3]
```

## Двусвязный список (Doubly Linked List)
Каждый узел ссылается и на следующий, и на предыдущий. Позволяет удалять/вставлять в обе стороны за `O(1)`.
Основа для **очереди** и **дека**.


## Стек (Stack)

**LIFO** — Last In, First Out. Как стопка тарелок: добавляем и забираем сверху.

Применения: вызовы функций, undo/redo, проверка скобок, обход в глубину.


```python
# Стек на обычном list — все операции O(1)
stack = []
stack.append(1)   # push
stack.append(2)
stack.append(3)
print(stack[-1])  # peek: 3
print(stack.pop()) # 3
print(stack)      # [1, 2]
```

### Классическая задача: проверка сбалансированности скобок


```python
def is_balanced(s):
    pairs = {')': '(', ']': '[', '}': '{'}
    stack = []
    for ch in s:
        if ch in '([{':
            stack.append(ch)
        elif ch in ')]}':
            if not stack or stack.pop() != pairs[ch]:
                return False
    return not stack

print(is_balanced('({[]})'))   # True
print(is_balanced('({[})'))    # False
```

## Очередь (Queue)

**FIFO** — First In, First Out. Кто пришёл первым, тот и обслужится первым.

На `list` нельзя делать `pop(0)` — это `O(n)`. Нужен **двусвязный список** или `collections.deque`.


```python
from collections import deque

queue = deque()
queue.append(1)        # enqueue, O(1)
queue.append(2)
queue.append(3)
print(queue.popleft()) # dequeue, O(1) → 1
print(queue)           # deque([2, 3])
```

### Очередь на двусвязном списке (реализация руками)


```python
class _Node:
    def __init__(self, value):
        self.value = value
        self.prev = None
        self.next = None

class Queue:
    def __init__(self):
        self.head = None
        self.tail = None

    def enqueue(self, value):       # O(1)
        node = _Node(value)
        if self.tail is None:
            self.head = self.tail = node
        else:
            node.prev = self.tail
            self.tail.next = node
            self.tail = node

    def dequeue(self):              # O(1)
        if self.head is None:
            raise IndexError('queue is empty')
        node = self.head
        self.head = node.next
        if self.head is None:
            self.tail = None
        else:
            self.head.prev = None
        return node.value

q = Queue()
for x in [1, 2, 3]:
    q.enqueue(x)
print(q.dequeue(), q.dequeue(), q.dequeue())  # 1 2 3
```

## Дек (Deque, Double-Ended Queue)

Можно вставлять и удалять с обоих концов за `O(1)`. В Python — `collections.deque`.
Внутри Python deque реализован как двусвязный список из блоков фиксированного размера.


```python
from collections import deque

d = deque([1, 2, 3])
d.appendleft(0)   # [0, 1, 2, 3]
d.append(4)       # [0, 1, 2, 3, 4]
d.popleft()       # 0
d.pop()           # 4
print(d)          # deque([1, 2, 3])
```

### Бонус: развернуть односвязный список


```python
def reverse(head):
    prev = None
    cur = head
    while cur:
        nxt = cur.next
        cur.next = prev
        prev = cur
        cur = nxt
    return prev
```

### Бонус: обнаружение цикла (черепаха и заяц)


```python
def has_cycle(head):
    slow = fast = head
    while fast and fast.next:
        slow = slow.next
        fast = fast.next.next
        if slow is fast:
            return True
    return False
```
