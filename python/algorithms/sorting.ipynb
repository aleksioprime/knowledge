# Сортировки


На собеседовании обычно спрашивают:
- Что такое **стабильность** сортировки
- Сложность по времени и памяти
- Какие сортировки `O(n²)`, какие `O(n log n)`, какие не сравнивают элементы

**Стабильная сортировка** сохраняет относительный порядок равных элементов.


## Сравнительная таблица

| Алгоритм | Время (avg) | Время (worst) | Память | Стабильна? |
|---|---|---|---|---|
| Bubble sort | O(n²) | O(n²) | O(1) | да |
| Selection sort | O(n²) | O(n²) | O(1) | нет |
| Insertion sort | O(n²) | O(n²) | O(1) | да |
| Merge sort | O(n log n) | O(n log n) | O(n) | да |
| Quick sort | O(n log n) | O(n²) | O(log n) | нет |
| Heap sort | O(n log n) | O(n log n) | O(1) | нет |
| Timsort (Python) | O(n log n) | O(n log n) | O(n) | да |


## Bubble sort — пузырьковая
Идея: проходим по массиву, попарно меняем местами соседей.


```python
def bubble_sort(arr):
    arr = arr.copy()
    n = len(arr)
    for i in range(n):
        # после i-го прохода последние i элементов уже на месте
        for j in range(n - 1 - i):
            if arr[j] > arr[j + 1]:
                arr[j], arr[j + 1] = arr[j + 1], arr[j]
    return arr

bubble_sort([64, 25, 12, 22, 11])
```

## Selection sort — выбором
На каждом шаге находим минимум в оставшейся части и ставим в начало.


```python
def selection_sort(arr):
    arr = arr.copy()
    n = len(arr)
    for i in range(n):
        min_idx = i
        for j in range(i + 1, n):
            if arr[j] < arr[min_idx]:
                min_idx = j
        arr[i], arr[min_idx] = arr[min_idx], arr[i]
    return arr

selection_sort([64, 25, 12, 22, 11])
```

## Insertion sort — вставками
Как карты в руке: берём следующий элемент и вставляем в нужное место среди уже отсортированных.

Лучшая `O(n²)` сортировка для **почти отсортированных** массивов: best case `O(n)`.


```python
def insertion_sort(arr):
    arr = arr.copy()
    for i in range(1, len(arr)):
        key = arr[i]
        j = i - 1
        # сдвигаем элементы, пока не найдём место
        while j >= 0 and arr[j] > key:
            arr[j + 1] = arr[j]
            j -= 1
        arr[j + 1] = key
    return arr

insertion_sort([64, 25, 12, 22, 11])
```

## Merge sort — сортировка слиянием
Divide & conquer: делим массив пополам, сортируем каждую половину рекурсивно, потом сливаем.

**Стабильна**, гарантированно `O(n log n)`, но нужна доп. память `O(n)`.


```python
def merge_sort(arr):
    if len(arr) <= 1:
        return arr
    mid = len(arr) // 2
    left = merge_sort(arr[:mid])
    right = merge_sort(arr[mid:])
    return merge(left, right)

def merge(left, right):
    result = []
    i = j = 0
    while i < len(left) and j < len(right):
        if left[i] <= right[j]:
            result.append(left[i]); i += 1
        else:
            result.append(right[j]); j += 1
    result.extend(left[i:])
    result.extend(right[j:])
    return result

merge_sort([64, 25, 12, 22, 11, 90, 33])
```

## Quick sort — быстрая
Тоже divide & conquer: выбираем **pivot**, делим массив на «меньше pivot» и «больше pivot», рекурсивно сортируем обе части.

В среднем `O(n log n)`. На уже отсортированных данных с плохим выбором pivot — `O(n²)`.


```python
def quick_sort(arr):
    if len(arr) <= 1:
        return arr
    pivot = arr[len(arr) // 2]   # берём средний — чуть надёжнее, чем крайний
    left  = [x for x in arr if x < pivot]
    mid   = [x for x in arr if x == pivot]
    right = [x for x in arr if x > pivot]
    return quick_sort(left) + mid + quick_sort(right)

quick_sort([64, 25, 12, 22, 11, 90, 33])
```

### Quick sort in-place (по памяти `O(log n)`)
Тот же алгоритм, но без выделения новых списков:


```python
def quick_sort_inplace(arr, low=0, high=None):
    if high is None:
        high = len(arr) - 1
    if low < high:
        p = partition(arr, low, high)
        quick_sort_inplace(arr, low, p - 1)
        quick_sort_inplace(arr, p + 1, high)
    return arr

def partition(arr, low, high):
    pivot = arr[high]
    i = low - 1
    for j in range(low, high):
        if arr[j] <= pivot:
            i += 1
            arr[i], arr[j] = arr[j], arr[i]
    arr[i + 1], arr[high] = arr[high], arr[i + 1]
    return i + 1

quick_sort_inplace([64, 25, 12, 22, 11, 90, 33])
```

## Timsort — что под капотом у Python
Встроенный `sorted()` и `list.sort()` используют **Timsort** — гибрид merge sort и insertion sort.

Свойства:
- стабилен;
- адаптивен (использует уже отсортированные участки → может быть `O(n)` на почти отсортированных);
- в худшем случае `O(n log n)`.


```python
data = [{'name': 'Alice', 'age': 30}, {'name': 'Bob', 'age': 25}, {'name': 'Cara', 'age': 30}]

# сортируем по ключу
sorted(data, key=lambda x: x['age'])

# по нескольким полям
sorted(data, key=lambda x: (x['age'], x['name']))
```

## Что спрашивают на собеседовании

- В чём разница между **stable** и **unstable** сортировкой?
- Какая сортировка лучше для **почти отсортированных** данных? *(Insertion / Timsort)*
- Можно ли отсортировать **за O(n)**? *(Можно — counting/radix/bucket sort, но они не сравнивают элементы и работают на ограниченных типах данных.)*
- Где использовать **merge sort**, а где **quick sort**? *(Merge — когда важна стабильность и есть память. Quick — когда память дорога.)*
