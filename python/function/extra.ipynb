# Модуль collections

## defaultdict

Удобный способ создания словаря с заданным значением по умолчанию для любого ключа, который еще не был добавлен в словарь

```python
from collections import defaultdict
d = defaultdict(int)
print(d['apple'])

d = defaultdict(list)
print(d['apple'])

d = defaultdict(set)
print(d['apple'])
```

## counter

Удобный способ подсчета количества встречаемых элементов в списке или другом итерируемом объекте. Он возвращает объект, который можно использовать как словарь, где ключами являются элементы, а значения – количество их вхождений.

```python
from collections import Counter
cnt = Counter(['red', 'blue', 'red', 'green', 'blue', 'blue'])
print(cnt)
print(dict(cnt))
```

## namedtuple

Можно создать именованный кортеж с заданными полями, что может быть удобно для работы с данными, которые имеют структуру, но не требуют создания класса.

```python
from collections import namedtuple

Point = namedtuple("Point", "x y")
print(issubclass(Point, tuple))

point = Point(2, 4)
print(point)

print(point.x)
print(point.y)

print(point[0])
print(point[1])
```

## Модуль itertools

## chain

Позволяет конкатенировать несколько итерируемых объектов в единый итератор.

```python
from itertools import chain

chained = chain('ab', [33])
print(next(chained))
print(next(chained))
print(next(chained))

for i in chain('1', [1, 2, 3], {3, 4}):
    print(i)
```

## groupby

Позволяет группировать элементы итерируемого объекта по заданному ключу.

```python
from itertools import groupby

grouper = lambda item: item['country']

data = [
    {'city': 'Москва', 'country': 'Россия'},
    {'city': 'Новосибирск', 'country': 'Россия'},
    {'city': 'Пекин', 'country': 'Китай'},
]

for key, group in groupby(data, key=grouper):
    print(key, *group)
```

## combinations и permutations

Генерируют все различные комбинации или перестановки элементов из заданного множества.

```python
from itertools import combinations

print(list(combinations('123', 2)))
```

```python
from itertools import permutations

print(list(permutations('123', 2)))
```
