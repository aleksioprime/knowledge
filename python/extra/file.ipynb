## Работа с JSON

```python
import json

# Преобразовать словарь → JSON-строку
data = {"name": "Alice", "age": 25}
json_str = json.dumps(data)
print(json_str)

# Преобразовать JSON-строку → словарь
json_str = '{"name": "Alice", "age": 25}'
data = json.loads(json_str)
print(data["name"])

# Записать JSON в файл
data = {"city": "Moscow", "temp": -5}
with open("data.json", "w", encoding="utf-8") as f:
    json.dump(data, f, ensure_ascii=False, indent=2)

# Прочитать JSON из файла
with open("data.json", "r", encoding="utf-8") as f:
    data = json.load(f)
print(data["city"])
```

```python
import json

# Список словарей (например, пользователи)
users = [
    {"id": 1, "name": "Alice", "age": 25},
    {"id": 2, "name": "Bob", "age": 30},
    {"id": 3, "name": "Eve", "age": 22}
]

# Сохранить в JSON-файл
with open("users.json", "w", encoding="utf-8") as f:
    json.dump(users, f, ensure_ascii=False, indent=2)

# Прочитать обратно
with open("users.json", "r", encoding="utf-8") as f:
    loaded_users = json.load(f)

print(loaded_users[1]["name"])  # Bob
```
