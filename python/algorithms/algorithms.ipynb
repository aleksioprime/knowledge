# Алгоритмы

BigO показывает как растёт функция с ростом входных данных или кол-ва операций:
- константная O(1) - операция над отдельными данными без циклов
- логарифмическая O(log n) - исходный объём данных делится на 2
- линейная O(n) - обработка массива в цикле или рекурсия
- квадратичная O(n2) - цикл в цикле
- факториальная O(!n)

Эффективность алгоритма оценивается:
- по времени (кол-во операций)
- по затраченным ресурсам (память)

Сложность операций с массивом:
- Получение элемента - O(1)
- Вставка в конец - O(1) при выделении памяти с запасом (амортизированна вставка) или O(n)
- Вставка в начало - O(n) (нужно копирование массива)
- Удаление с начала/по индексу - O(n) (нужно копирование массива)
- Удаление с конца - O(1)

## Линейный поиск

```python
arr = [4, 7, 1, 9, 3]
target = 9

for x in arr:
    if x == target:
        print("Найден")
```

## Бинарный поиск

```python
arr = [1, 3, 4, 7, 9]
target = 9

left, right = 0, len(arr) - 1

while left <= right:
    mid = (left + right) // 2
    if arr[mid] == target:
        print("Найден")
        break
    elif arr[mid] < target:
        left = mid + 1
    else:
        right = mid - 1
```

## Рекурсия

```python
def factorial(n):
    if n == 0:
        return 1
    else:
        return n * factorial(n - 1)

print(factorial(5))  # Вывод: 120
```

## Сортировка

```python
# Сортировка пузырьком

arr = [64, 25, 12, 22, 11]

n = len(arr)

for i in range(n):
    for j in range(0, n - 1):
        if arr[j] > arr[j + 1]:
            arr[j], arr[j + 1] = arr[j + 1], arr[j]

print(arr)
```

```python
# Сортировка выбором

arr = [64, 25, 12, 22, 11]

n = len(arr)

for i in range(n):
    min_index = i
    for j in range(i + 1, n):
        if arr[j] < arr[min_index]:
            min_index = j

    arr[i], arr[min_index] = arr[min_index], arr[i]

print(arr)
```

```python
# Быстрая сортировка

def quick_sort(arr):

    # если массив пустой или содержит один элемент, он уже отсортирован
    if len(arr) <= 1:
        return arr

    # выбираем опорный элемент (здесь первый элемент массива)
    pivot = arr[0]

    # разделяем массив на элементы меньше и больше опорного
    left = [x for x in arr[1:] if x <= pivot]
    right = [x for x in arr[1:] if x > pivot]

    # рекурсивно сортируем подмассивы и объединяем результаты
    return quick_sort(left) + [pivot] + quick_sort(right)


arr = [64, 25, 12, 22, 11]
print(quick_sort(arr))
```

```python
def quick_sort(arr, low, high):
    """Реализация быстрой сортировки"""

    # Если есть хотя бы два элемента для сортировки
    if low < high:
        # Находим индекс опорного элемента
        p = partition(arr, low, high)

        # Рекурсивно сортируем элементы до и после разделения
        quick_sort(arr, low, p - 1)
        quick_sort(arr, p + 1, high)


def partition(arr, low, high):
    """Разделяет массив на две части и возвращает индекс опорного элемента"""

    # Выбираем опорный элемент (здесь последний элемент массива)
    pivot = arr[high]
    # Индекс меньшего элемента
    i = low - 1

    # Проходим по всем элементам массива
    for j in range(low, high):
        # Если текущий элемент меньше или равен опорному, меняем его местами с элементом на позиции i + 1
        if arr[j] <= pivot:
            i += 1
            arr[i], arr[j] = arr[j], arr[i]

    # Меняем опорный элемент местами с элементом на позиции i + 1
    arr[i + 1], arr[high] = arr[high], arr[i + 1]
    return i + 1


arr = [64, 25, 12, 22, 11]
quick_sort(arr, 0, len(arr) - 1)
print(arr)
```

## Связанный список

Связанный список — это структура данных, состоящая из узлов (nodes), где каждый узел:
- хранит значение
- содержит ссылку на следующий узел (а иногда и на предыдущий)

Сложность:
- вставка в начало/конец O(1), в середину - поиск O(n) и вставка O(1)
- удаление в начало/конец O(1), в середину - поиск O(n) и вставка O(1)
- получение из начала/конца O(1), из середины - O(n)

```python
class Node:
    """ Класс для представления узла связного списка """
    def __init__(self, value):
        self.value = value
        self.next = None

class LinkedList:
    """ Класс для представления связного списка """
    def __init__(self):
        self.head = None

    def append(self, value):
        """ Добавляет новый узел с заданным значением в конец связного списка """
        node = Node(value)

        # Если список пуст, новый узел становится головой
        if self.head is None:
            self.head = node
            return

        # Иначе, проходим до конца списка и добавляем новый узел
        current = self.head
        while current.next:
            current = current.next

        current.next = node

    def prepend(self, value):
        """ Добавляет новый узел с заданным значением в начало связного списка """
        node = Node(value)
        node.next = self.head
        self.head = node

    def remove(self, value):
        """ Удаляет первый узел с заданным значением из связного списка """
        # Если список пуст, ничего не делаем
        if self.head is None:
            return

        # Если голова содержит значение для удаления
        if self.head.value == value:
            self.head = self.head.next
            return

        # Иначе, ищем узел для удаления
        current = self.head
        while current.next:
            if current.next.value == value:
                current.next = current.next.next
                return
            current = current.next

    def find(self, value):
        """ Ищет узел с заданным значением в связном списке """
        current = self.head
        # Ищем узел с заданным значением
        while current:
            if current.value == value:
                return current
            current = current.next
        return None
```

## Очередь (Queue)

Принцип FIFO: First In, First Out - первым вошёл, первым вышел

Оптимальнее делать на двусвязном списке (в списке удаление первого элемента занимает O(n))

```python
class Node:
    def __init__(self, value):
        self.value = value
        self.prev = None
        self.next = None


class DequeQueue:
    """Очередь FIFO на двусвязном списке: добавляем в tail, забираем из head."""

    def __init__(self):
        self.head = None
        self.tail = None

    def enqueue(self, value):
        """Добавить элемент в конец очереди. O(1)"""
        node = Node(value)

        if self.tail is None:
            self.head = self.tail = node
        else:
            node.prev = self.tail
            self.tail.next = node
            self.tail = node

    def dequeue(self):
        """Удалить и вернуть элемент из начала очереди. O(1)"""
        if self.head is None:
            raise IndexError("Dequeue from empty queue")

        node = self.head
        new_head = node.next

        if new_head is None:
            self.head = self.tail = None
        else:
            new_head.prev = None
            self.head = new_head

        return node.value
```

## Поиск в ширину (BFS)

Сложность O(v+e)

```python
graph = {
    'A': ['B', 'C'],
    'B': ['D', 'E'],
    'C': ['F'],
    'D': [],
    'E': [],
    'F': []
}

def bfs(graph, start):
    """Обход графа в ширину (BFS)"""
    # словарь посещённых вершин
    visited = {}
    # обычный список как очередь
    queue = []

    queue.append(start)
    visited[start] = True

    while len(queue) > 0:
        # взять первый элемент
        vertex = queue[0]
        # удалить его из очереди
        queue = queue[1:]

        # обработать вершину
        print(vertex, end=" ")

        # добавить соседей в очередь, если они не посещены
        for neighbor in graph[vertex]:
            if neighbor not in visited:
                visited[neighbor] = True
                queue.append(neighbor)

# BFS посещает только достижимые вершины от стартовой точки
bfs(graph, 'B')
```

## Поиск в глубину (DFS)

```python
graph = {
    'A': ['B', 'C'],
    'B': ['D', 'E'],
    'C': ['F'],
    'D': [],
    'E': [],
    'F': []
}

def dfs(graph, start):
    """Обход графа в глубину (DFS)"""
    # словарь посещённых вершин
    visited = {}
    # обычный список как стек
    stack = []

    # добавляем стартовую вершину в стек
    stack.append(start)

    while len(stack) > 0:
        # взять верхний элемент
        vertex = stack.pop()

        if vertex not in visited:
            # отмечаем вершину как посещённую
            visited[vertex] = True
            print(vertex, end=" ")

            # добавляем соседей в стек в обратном порядке, если они не посещены
            neighbors = graph[vertex]
            for i in range(len(neighbors) - 1, -1, -1):
                neighbor = neighbors[i]
                if neighbor not in visited:
                    stack.append(neighbor)

def dfs_recursive(graph, start, visited=None):
    """Обход графа в глубину (DFS) рекурсивно"""

    # инициализация словаря посещённых вершин при первом вызове
    if visited is None:
        visited = {}

    visited[start] = True
    print(start, end=" ")

    # рекурсивно посещаем соседей вершины
    for neighbor in graph[start]:
        if neighbor not in visited:
            dfs_recursive(graph, neighbor, visited)

dfs_recursive(graph, 'B')
```
