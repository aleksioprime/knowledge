### Байты

В Python объект bytes представляет собой последовательность целых чисел от 0 до 255 (один байт). При выводе Python может показывать его в шестнадцатеричном виде.

```python
# Байты в Python являются неизменяемыми (immutable)
b = b'abc'
b[0] = 65
```

```python
# Массив байтов (bytearray) является изменяемым (mutable)
ba = bytearray(b'abc')
ba[0] = 65
print(ba)
```

### Кодирование и декодирование

Стандарт Unicode чётко разделяет:
- Идентификатор символа (кодовую позицию) – это число от 0 до 1 114 111 (по основанию 10), которое в стандарте Unicode записывается шестнадцатеричными цифрами (в количестве от 4 до 6) с префиксом «U+»
- Байтовое представление символа - это последовательность одного или нескольких байтов, получаемая в результате применения конкретной кодировки к кодовой позиции символа.

Кодировка — это алгоритм преобразования кодовых позиций Unicode в последовательность байтов и обратно. Один и тот же символ может иметь разное байтовое представление в разных кодировках.

Если байт соответствует печатному ASCII-символу (коды 32–126), выводится сам символ. В остальных случаях используется escape-последовательность вида \xHH, где HH — двухзначное шестнадцатеричное представление байта.

```python
s = 'café'
print(f"Text = {s}, len(s) = {len(s)}, type(s) = {type(s)}")
b = s.encode('utf8') # кодируем строку в байты (str -> bytes)
print(f"Bytes = {b}, len(b) = {len(b)}, type(b) = {type(b)}")
c = b.decode('utf8') # декодируем обратно в строку (bytes -> str)
print(f"Decoded = {c}, len(c) = {len(c)}, type(c) = {type(c)}")
```
