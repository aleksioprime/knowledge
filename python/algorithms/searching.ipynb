# Поиск в массиве


Самая частая задача на собеседовании. Два алгоритма должны быть в голове наизусть: линейный поиск и бинарный.


## Линейный поиск — O(n)
Перебираем все элементы подряд. Работает на любых данных.


```python
def linear_search(arr, target):
    for i, x in enumerate(arr):
        if x == target:
            return i
    return -1

linear_search([4, 7, 1, 9, 3], 9)   # 3
```

## Бинарный поиск — O(log n)
Работает **только на отсортированном массиве**. На каждом шаге отбрасываем половину.

Идея: смотрим середину, сравниваем с target и идём в нужную половину.


```python
def binary_search(arr, target):
    left, right = 0, len(arr) - 1
    while left <= right:
        mid = (left + right) // 2
        if arr[mid] == target:
            return mid
        elif arr[mid] < target:
            left = mid + 1
        else:
            right = mid - 1
    return -1

binary_search([1, 3, 4, 7, 9, 11, 15], 9)   # 4
```

### Подводные камни
- `mid = (left + right) // 2` — в Python нет переполнения, всё ок.
  В C/Java надо `mid = left + (right - left) // 2` чтобы избежать overflow.
- Не забывать `left <= right` (со знаком равно).
- Сужать диапазон обязательно через `mid + 1` / `mid - 1`, иначе бесконечный цикл.


## Бинарный поиск рекурсивно


```python
def binary_search_rec(arr, target, left=0, right=None):
    if right is None:
        right = len(arr) - 1
    if left > right:
        return -1
    mid = (left + right) // 2
    if arr[mid] == target:
        return mid
    if arr[mid] < target:
        return binary_search_rec(arr, target, mid + 1, right)
    return binary_search_rec(arr, target, left, mid - 1)

binary_search_rec([1, 3, 4, 7, 9, 11, 15], 11)
```

## Модуль `bisect` — стандартный бинарный поиск Python
Удобнее писать в проде, чем своими руками.


```python
import bisect

arr = [1, 3, 4, 7, 9, 11, 15]

# индекс, куда вставить, чтобы сохранить сортировку
print(bisect.bisect_left(arr, 7))   # 3 (позиция самого 7)
print(bisect.bisect_right(arr, 7))  # 4 (позиция за 7)

# вставка с сохранением порядка
bisect.insort(arr, 8)
print(arr)   # [1, 3, 4, 7, 8, 9, 11, 15]
```

## Когда что использовать

- **Неотсортированный массив** → линейный (либо отсортировать, если поиск будет частым).
- **Отсортированный массив** → бинарный (`bisect`).
- **Поиск по ключу с равными значениями** → хеш-таблица `dict` (`O(1)`).
- **Поиск ближайшего значения** → бинарный (`bisect` подскажет позицию).
