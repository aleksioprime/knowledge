# Рекурсия


**Рекурсия** — функция, которая вызывает сама себя.

Обязательны два элемента:
1. **База** — условие выхода (иначе бесконечный вызов и `RecursionError`).
2. **Шаг** — вызов с уменьшенными данными, ведущий к базе.


## Факториал


```python
def factorial(n):
    if n <= 1:              # база
        return 1
    return n * factorial(n - 1)   # шаг

factorial(5)   # 120
```

## Числа Фибоначчи: наивная рекурсия — O(2ⁿ)
Один и тот же `fib(k)` пересчитывается много раз → экспоненциальная сложность.


```python
def fib(n):
    if n < 2:
        return n
    return fib(n - 1) + fib(n - 2)

fib(10)   # 55
```

## Мемоизация — O(n)
Кэшируем результат для каждого `n`. Готовый декоратор `functools.lru_cache`:


```python
from functools import lru_cache

@lru_cache(maxsize=None)
def fib(n):
    if n < 2:
        return n
    return fib(n - 1) + fib(n - 2)

fib(100)   # мгновенно, без кэша было бы вечность
```

## Лимит рекурсии в Python
По умолчанию ~1000. Python не оптимизирует **хвостовую рекурсию** — слишком глубокая рекурсия падает с `RecursionError`.


```python
import sys
print(sys.getrecursionlimit())   # 1000
# sys.setrecursionlimit(10_000)  # можно повысить, но лучше переписать на цикл
```

## Когда рекурсия — естественный выбор

- Структуры с самоповторением: деревья, графы, JSON, AST.
- Алгоритмы divide & conquer: merge sort, quick sort, бинарный поиск.
- Backtracking: перебор с возвратами (перестановки, расстановка ферзей, sudoku).


## Backtracking: все перестановки


```python
def permutations(arr):
    result = []
    def backtrack(path, remaining):
        if not remaining:
            result.append(path)
            return
        for i, x in enumerate(remaining):
            backtrack(path + [x], remaining[:i] + remaining[i+1:])
    backtrack([], arr)
    return result

permutations([1, 2, 3])
```

## Backtracking: задача о N ферзях
Расставить N ферзей на доске N×N так, чтобы они друг друга не били.


```python
def solve_n_queens(n):
    result = []
    cols = set()
    diag1 = set()   # row - col
    diag2 = set()   # row + col
    board = [-1] * n

    def backtrack(row):
        if row == n:
            result.append(board.copy())
            return
        for col in range(n):
            if col in cols or (row - col) in diag1 or (row + col) in diag2:
                continue
            cols.add(col); diag1.add(row - col); diag2.add(row + col)
            board[row] = col
            backtrack(row + 1)
            cols.remove(col); diag1.remove(row - col); diag2.remove(row + col)

    backtrack(0)
    return result

print(f'Решений для N=4: {len(solve_n_queens(4))}')   # 2
print(f'Решений для N=8: {len(solve_n_queens(8))}')   # 92
```

## Как переводить рекурсию в цикл
Любую рекурсию можно переписать через явный стек (`list`). Иногда это нужно из-за лимита Python.


```python
# Факториал через цикл
def factorial_iter(n):
    result = 1
    for i in range(2, n + 1):
        result *= i
    return result

factorial_iter(5)
```
