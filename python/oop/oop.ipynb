## Использование метода объекта

```python
class Person:

    def __init__(self, name):
        self.name = name

    def say_hi(self):
        # Метод объекта (вызвать можно только из объекта)
        # self - ссылка на текущий объект
        print(f"Hi, my name is {self.name}")

p = Person("Mike")
p.say_hi()
```

## Использование метода класса

```python
class Person:
    workplace = "My Company"

    @classmethod
    def get_my_workplace(cls):
        # Метод класса (вызывать можно без создания объекта)
        # cls - ссылка на текущий класс
        return cls.workplace

print(Person.get_my_workplace())

p = Person()
print(p.get_my_workplace())
```

## Использование статического метода

```python
from datetime import datetime

class A:
    @staticmethod
    def get_current_datetime():
        # статические методы не принимают self или cls
        # вызываются из класса или объекта
            return datetime.now()

print(A.get_current_datetime())

a = A()
print(a.get_current_datetime())
```

## Декоратор property

```python
class Person:
    first_name = str
    last_name = str

    def __init__(self, first_name: str, last_name: str):
        self.first_name = first_name
        self.last_name = last_name

    # Динамический атрибут (геттер)
    @property
    def full_name(self):
        return f"{self.first_name} {self.last_name}"

    # Сеттер (для установки нового значения атрибутов с валидацией)
    @full_name.setter
    def full_name(self, value):
        name = value.split()
        self.first_name = name[0]
        self.last_name = name[1]

p = Person("Mike", "Jagger")
print(p.full_name)

p.full_name = "John Walker"
print(p.full_name)
```

## Абстрактный класс

```python
## Из абстрактого класса нельзя создать объект, он определяет обязательный интерфейс для классов-потомков

from abc import ABC, abstractmethod

class Animal(ABC):

    @abstractmethod
    def say_something(self):
        pass

# Нельзя создать объект
# a = Animal()

class Cat(Animal):

    def say_something(self):
        print("Meaw")

c = Cat()
```

## Разница между `__new__` и `__init__`

```python
class A:
    def __new__(cls, *args, **kwargs):
        # Создаёт объект и возвращает его
        print("New")
        obj = super().__new__(cls, *args, **kwargs)
        return obj

    def __init__(self):
        # Получает готовый объект и заполняет его атрибутами
        print("Init")
        self.some = 1

a = A()
```

## Разница между `__str__` и `__repr__`

```python
class Cat:
    name: str

    def __init__(self, name):
        self.name = name

    def __repr__(self):
        # Формальное техническое представление объекта
        return f"Cat(name='{self.name}')"

    def __str__(self):
        # Представление объекта для пользователя
        return f"It is my cat {self.name}"

cat = Cat("Barsik")
print(cat)
print(repr(cat))

# Если нет __str__, то используется __repr__. Если нет __repr__, то берётся стандартное адресное представление объекта
```

## Использование функции `super()`

```python
class A:
    def __init__(self, a):
        self.a = a

class B(A):
    def __init__(self, a, b):
        # Обращение к методу родительского класса и
        # передача туда аргумента a для инициализации атрибута a
        super().__init__(a)
        self.b = b
```

## Инкапсуляция

```python
class BankAccount:
    # Объединение переменных и функций одной цели в одном классе
    def __init__(self, balance):
        self.__balance = balance

    # Разграничение абстраций от реализаций (сокрытие реализаций)
    @property
    def balance(self):
        return self.__balance

    @balance.setter
    def balance(self, value):
        if value >= 0:
            self.__balance = value
        else:
            print("Баланс не может быть отрицательным")

account = BankAccount(1000)

print(account.balance)
account.balance = 2000
print(account.balance)

account.balance = -100
```

# Полиморфизм

```python
class A:
    def hi(self):
        print(f"Hi from A")

class B:
    def hi(self):
        print(f"Hi from B")

# Параметрический полиморфизм
def some(obj):
    # Функция выполняет заданный метод объектов, которые могут относиться к разным классам
    obj.hi()

a = A()
some(a)

b = B()
some(b)

# Ad-hoc полиморфизм (одна и та же функция ведёт себя по-разному в зависимости от типа аргумента)
def to_json(value):
    if isinstance(value, int):
        return str(value)
    if isinstance(value, float):
        return str(value)
    if isinstance(value, str):
        return f'"{value}"'
    if isinstance(value, list):
        return '[' + ','.join(to_json(x) for x in value) + ']'

# Ad-hoc полиморфизм (перегрузка операторов и функций)
class Vector:
    def __init__(self, x, y):
        self.x, self.y = x, y

    def __add__(self, other):
        return Vector(self.x + other.x, self.y + other.y)

v1 = Vector(1, 2)
v2 = Vector(3, 4)
v3 = v1 + v2   # работает благодаря перегрузке
print(v3.x, v3.y)  # 4 6
```

## Метаклассы
Метакласс — это “класс для классов”. Он управляет созданием самих классов (например, чтобы проверить, соответствуют ли классы определённым требованиям)

```python
# Определяем метакласс
class MyMeta(type):
    def __new__(cls, name, bases, attrs):
        print(f"Создаётся класс: {name}")
        return super().__new__(cls, name, bases, attrs)

# Используем его
class MyClass(metaclass=MyMeta):
    pass
```
