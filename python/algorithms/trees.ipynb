# Деревья


**Дерево** — иерархическая структура из узлов: один **корень** (root), у каждого узла — несколько **детей**, у каждого узла (кроме корня) — один **родитель**.

Без циклов. Терминология: лист (leaf), глубина, высота, поддерево.


## Бинарное дерево
У каждого узла **не более двух детей**: `left` и `right`.


```python
class TreeNode:
    def __init__(self, value, left=None, right=None):
        self.value = value
        self.left = left
        self.right = right

# дерево:
#         1
#        / \
#       2   3
#      / \
#     4   5

root = TreeNode(1,
                TreeNode(2, TreeNode(4), TreeNode(5)),
                TreeNode(3))
```

## Обходы дерева

Все обходы — `O(n)`. Запомнить порядок легко: pre/in/post — где обрабатывается **узел**, относительно `left`/`right`.


### Pre-order (NLR): узел → левое → правое


```python
def preorder(node):
    if node is None: return
    print(node.value, end=' ')
    preorder(node.left)
    preorder(node.right)

preorder(root)   # 1 2 4 5 3
```

### In-order (LNR): левое → узел → правое
Для BST даёт значения **в отсортированном порядке**.


```python
def inorder(node):
    if node is None: return
    inorder(node.left)
    print(node.value, end=' ')
    inorder(node.right)

inorder(root)   # 4 2 5 1 3
```

### Post-order (LRN): левое → правое → узел
Удобно для удаления дерева — сначала освобождаем детей.


```python
def postorder(node):
    if node is None: return
    postorder(node.left)
    postorder(node.right)
    print(node.value, end=' ')

postorder(root)   # 4 5 2 3 1
```

### Level-order (BFS) — по уровням


```python
from collections import deque

def level_order(root):
    if not root: return
    q = deque([root])
    while q:
        node = q.popleft()
        print(node.value, end=' ')
        if node.left:  q.append(node.left)
        if node.right: q.append(node.right)

level_order(root)   # 1 2 3 4 5
```

## BST — Binary Search Tree

Бинарное дерево со свойством:
- слева от узла — значения **меньше**
- справа — **больше**

Это даёт **поиск, вставку, удаление за `O(log n)`** — если дерево сбалансировано.
Если несбалансировано (например, вставляли отсортированные данные) — деградирует до `O(n)`.

Производственные реализации (`std::map` в C++, TreeMap в Java) — **самобалансирующиеся**: AVL, Red-Black tree.


```python
class BST:
    def __init__(self):
        self.root = None

    def insert(self, value):
        self.root = self._insert(self.root, value)

    def _insert(self, node, value):
        if node is None:
            return TreeNode(value)
        if value < node.value:
            node.left = self._insert(node.left, value)
        elif value > node.value:
            node.right = self._insert(node.right, value)
        return node

    def contains(self, value):
        node = self.root
        while node:
            if value == node.value: return True
            node = node.left if value < node.value else node.right
        return False

bst = BST()
for x in [5, 3, 7, 1, 4, 6, 8]:
    bst.insert(x)
print(bst.contains(4), bst.contains(100))   # True False
inorder(bst.root)   # 1 3 4 5 6 7 8  — отсортировано
```

## Куча (Heap, Binary Heap)

Бинарное дерево, в котором:
- **Min-heap**: значение в каждом узле ≤ значений детей. Корень — минимум.
- **Max-heap**: ≥. Корень — максимум.

Хранится **в массиве**: для узла `i` дети находятся на `2i+1` и `2i+2`.

Сложность:
- `push` — O(log n)
- `pop` (min/max) — O(log n)
- `peek` — O(1)

В Python — модуль `heapq` (min-heap).


```python
import heapq

h = []
for x in [5, 1, 3, 9, 2]:
    heapq.heappush(h, x)

print(h[0])                # peek минимума: 1
print(heapq.heappop(h))    # 1
print(heapq.heappop(h))    # 2
print(heapq.heappop(h))    # 3

# превратить готовый список в кучу — O(n)
arr = [5, 1, 3, 9, 2]
heapq.heapify(arr)
print(arr)   # [1, 2, 3, 9, 5]
```

### Max-heap через инверсию знака
Стандартного max-heap в Python нет — пушим со знаком минус.


```python
import heapq

h = []
for x in [5, 1, 3, 9, 2]:
    heapq.heappush(h, -x)

print(-heapq.heappop(h))   # 9 — максимум
```

### Применения кучи
- Top-K элементов: `heapq.nlargest(k, arr)` / `nsmallest`.
- Очередь с приоритетом (Priority Queue), например в Dijkstra.
- Heap sort.
- Медиана потока (через две кучи).
