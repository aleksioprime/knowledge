## Сценарий 6. Inline-зависимости в скрипте (PEP 723)

Можно описать зависимости прямо в shebang-блоке скрипта — uv создаст временный venv и запустит.


```python
script = '''# /// script
# requires-python = ">=3.12"
# dependencies = ["httpx", "rich"]
# ///

import httpx
from rich import print

r = httpx.get("https://httpbin.org/get")
print(r.json())
'''
with open('inline_demo.py', 'w') as f:
    f.write(script)
print('inline_demo.py created')
```

```python
# uv сам поставит httpx и rich в одноразовый venv
!uv run inline_demo.py 2>&1 | head -20
```
