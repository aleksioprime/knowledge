# Big O Notation


**Big O** описывает, как растёт время или память алгоритма с ростом размера входных данных `n`.

Мы интересуемся **поведением при `n → ∞`** — константы и младшие члены отбрасываем.

Если функция работает `3n² + 5n + 100` операций — пишем `O(n²)`.


## Шкала сложностей (от лучшей к худшей)

| Big O | Название | Пример |
|---|---|---|
| O(1) | константная | доступ к элементу массива по индексу |
| O(log n) | логарифмическая | бинарный поиск |
| O(n) | линейная | проход по массиву |
| O(n log n) | линейно-логарифмическая | merge sort, quick sort (средн.) |
| O(n²) | квадратичная | вложенный цикл, bubble sort |
| O(2ⁿ) | экспоненциальная | наивный Фибоначчи рекурсией |
| O(n!) | факториальная | перебор перестановок (TSP brute force) |


## Примеры для каждой сложности


### O(1) — константная
Доступ к элементу по индексу, hash-lookup.


```python
def first_element(arr):
    return arr[0]      # одна операция вне зависимости от длины

first_element([1, 2, 3, 4, 5])
```

### O(log n) — логарифмическая
На каждом шаге задача уменьшается в 2 раза.


```python
def binary_search(arr, target):
    left, right = 0, len(arr) - 1
    while left <= right:
        mid = (left + right) // 2
        if arr[mid] == target:
            return mid
        if arr[mid] < target:
            left = mid + 1
        else:
            right = mid - 1
    return -1

binary_search([1, 3, 5, 7, 9, 11], 7)
```

### O(n) — линейная
Один проход по данным.


```python
def find_max(arr):
    m = arr[0]
    for x in arr:            # n операций
        if x > m:
            m = x
    return m

find_max([3, 1, 4, 1, 5, 9, 2, 6])
```

### O(n log n) — линейно-логарифмическая
Любая эффективная сортировка сравнениями.


```python
# Встроенный sorted() — Timsort, O(n log n)
sorted([5, 3, 8, 1, 9, 2])
```

### O(n²) — квадратичная
Вложенный цикл по всем парам.


```python
def has_duplicate_pair(arr):
    n = len(arr)
    for i in range(n):
        for j in range(i + 1, n):   # ~ n² / 2 пар
            if arr[i] == arr[j]:
                return True
    return False

has_duplicate_pair([1, 2, 3, 2])
```

### O(2ⁿ) — экспоненциальная
Рекурсия с 2 вызовами на каждом шаге без мемоизации.


```python
def fib(n):
    if n < 2:
        return n
    return fib(n - 1) + fib(n - 2)   # 2 вызова → 2ⁿ

fib(10)
```

### O(n!) — факториальная
Перебор всех перестановок. На практике почти всегда нужно избегать.


```python
from itertools import permutations

list(permutations([1, 2, 3]))   # 3! = 6 вариантов
```

## Время и память
Big O оценивает не только **время**, но и **дополнительную память**.

Пример: merge sort — `O(n log n)` по времени, но `O(n)` по памяти (нужен буфер для слияния).
Quick sort — `O(n log n)` по времени, `O(log n)` по памяти (стек рекурсии).


## Worst / Average / Best case

- **Best case** — лучший возможный сценарий (бинарный поиск нашёл в середине: `O(1)`).
- **Average case** — типичная нагрузка.
- **Worst case** — гарантия сверху (то, что обычно пишут).

Quick sort: средн. `O(n log n)`, худш. `O(n²)` (если pivot всегда минимальный).


## Амортизированная сложность
Когда отдельная операция иногда дорогая, но в среднем по N операций — дешёвая.

Пример: `list.append` в Python. Обычно `O(1)`, но иногда массив надо перевыделить и скопировать → `O(n)`.
В среднем по N добавлений — амортизированно `O(1)`.


## Сложность операций встроенных структур Python

### list
| Операция | Сложность |
|---|---|
| `lst[i]` (индекс) | O(1) |
| `lst.append(x)` | O(1) амортизированно |
| `lst.insert(0, x)` (в начало) | O(n) |
| `lst.pop()` (с конца) | O(1) |
| `lst.pop(0)` (с начала) | O(n) |
| `x in lst` | O(n) |
| `len(lst)` | O(1) |

### dict / set
| Операция | Сложность |
|---|---|
| `d[k]`, `k in d` | O(1) средн., O(n) худш. |
| `d[k] = v` | O(1) средн. |
| `del d[k]` | O(1) средн. |

### collections.deque
| Операция | Сложность |
|---|---|
| `append`, `appendleft` | O(1) |
| `pop`, `popleft` | O(1) |
| `d[i]` (по индексу в середине) | O(n) |


## Как думать о сложности

1. **Считай вложенные циклы**: 2 вложенных по `n` → `O(n²)`.
2. **Деление пополам** → почти всегда логарифм.
3. **Hash-структуры** (`dict`, `set`) — твой друг для `O(1)` lookup.
4. **Сортировка** — почти бесплатно, `O(n log n)` через `sorted()`.
5. На собеседовании ВСЕГДА называй сложность по времени **и** по памяти.
