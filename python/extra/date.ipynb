## Работа с датой и временем

```python
from datetime import date, time, datetime, timedelta
```

```python
# Получение текущей даты и времени
now = datetime.now()      # текущие дата и время
today = date.today()      # только дата
utc_now = datetime.utcnow()  # время по UTC
```

```python
# Создание даты и времени вручную
d = date(2025, 11, 12)
t = time(14, 30, 0)
dt = datetime(2025, 11, 12, 14, 30)
```

```python
# Форматирование даты
now = datetime.now()
print(now.strftime("%Y-%m-%d %H:%M:%S"))
```

```python
# Преобразование строки в дату
s = "2025-11-12 14:30"
dt = datetime.strptime(s, "%Y-%m-%d %H:%M")
```

```python
import time
# Timestamp (временная метка) — это число, которое показывает,
# сколько секунд прошло с 1 января 1970 года, 00:00:00 UTC,
# то есть с момента, называемого Unix Epoch (“эпоха Unix”).

now = datetime.now()
ts = time.time()              # текущий timestamp (в секундах)
dt = datetime.fromtimestamp(ts)  # timestamp → datetime
```

```python
# Временные зоны

from datetime import datetime, timezone, timedelta

utc = datetime.now(timezone.utc)  # текущее UTC-время
moscow = utc.astimezone(timezone(timedelta(hours=3)))

print(moscow)
```
